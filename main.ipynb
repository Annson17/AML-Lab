```python
# Indexing/Slicing
s=np.arange(100)
s
```

```python
s=np.arange(6)**2
s
```

```python
s=np.arange(13)**2
s
```

```python
#Use bracket notation to get specific elements
s[0],s[4],s[-1]
```

```python
#Use : to indicate a range of values, array[start:stop:step] where start is inclusive and stop is exclusive
s[1:5]
```

```python
s[0:5:2] #start at index 0, stop at index 5, step by 2
```

```python
s[:3] #start at index 0, stop at index 3, step by 1
```

```python
r=np.arange(100)
r.resize((10,10)) #.resize() to change shape of array
r
```

```python
# Use bracket notation to slice: array[row,column] where row and column can be a single index or a range of indices
r[2,4] #get the value at row 2, column 4
```

```python
#2nd Example of slicing
r[0:5,0:5] #get the values from row 0 to 4 and column 0 to 4
```

```python
r[:3,:-1] #get the values from row 0 to 2 and column 0 to 8













```
