```python
"""3. Create a NumPy array containing the numbers from 10 to 50. Display:

The array
Number of elements
Shape
Dimension
Data type
"""

c=np.arange(10, 51)
print("Array:", c)
print("Number of elements:", c.size)
print("Shape:", c.shape)
print("Dimension:", c.ndim)
print("Data type:", c.dtype)
```

```python
"""4. Using NumPy, create:

A 1D array containing 10 zeros
A 3 × 3 array containing ones
A 4 × 4 identity matrix
An array containing numbers from 5 to 50 with a step size of 5.
""" 

d=np.zeros(10)
e=np.ones((3, 3))
f=np.eye(4)
g=np.arange(5, 51, 5)
d,e,f,g
```

```python
"""5. Create an array containing numbers from 1 to 24. Reshape it into:

4 × 6
6 × 4
2 × 3 × 4

Display the shape and ndim of each result."""

a=np.arange(1, 25)
b=a.reshape(4, 6)
c=b.reshape(6, 4)
d=c.reshape(2, 3, 4)
print("4 x 6 array:\n", b)
print("Shape:", b.shape, "Dimension:", b.ndim)
print("\n6 x 4 array:\n", c)
print("Shape:", c.shape, "Dimension:", c.ndim)
print("\n2 x 3 x 4 array:\n", d)
```

```python
"""6. Consider: marks = np.array([78, 85, 92, 67, 88, 73, 95, 81])

Write NumPy statements to display:

First element
Last element
Third element
First four elements
Last three elements"""

marks = np.array([78, 85, 92, 67, 88, 73, 95, 81])
print("First element:", marks[0])
print("Last element:", marks[-1])
print("Third element:", marks[2])
print("First four elements:", marks[:4])
print("Last three elements:", marks[-3:])
```

```python
"""7. Create:
a = np.array([10, 20, 30, 40, 50])
b = np.array([5, 10, 15, 20, 25])

Calculate:

a + b
a - b
a * b
a / b
Square of every element in a"""

a=np.array([10, 20, 30, 40, 50])
b=np.array([5, 10, 15, 20, 25])
print("a + b:", a + b)
print("a - b:", a - b)
print("a * b:", a * b)
print("a / b:", a / b)
print("Square of every element in a:", a ** 2)
```
